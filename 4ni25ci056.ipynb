```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(data=df, x="PetalLengthCm", bins=10)
plt.hist(df["PetalLengthCm"], bins=10)
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(df)

df.boxplot(column=df.columns[2])
plt.show()
```

```python
plt.scatter(df["PetalLengthCm"], df["PetalWidthCm"])
plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```

```python
import matplotlib.pyplot as plt

features = df.columns[:-1]
species = df["species"].unique()

fig, axes = plt.subplots(len(features), len(features), figsize=(10, 10))

for i, x in enumerate(features):
    for j, y in enumerate(features):
        ax = axes[i, j]

        for s in species:
            data = df[df["species"] == s]
            ax.scatter(data[x], data[y], label=s)

        ax.set_xlabel(x)
        ax.set_ylabel(y)

plt.tight_layout()
plt.show()
```

```python
correlation = df.select_dtypes("number").corr()
sns.heatmap(
correlation,
annot=True,
cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()
```
